```python
import time
import requests
import re
from openai import OpenAI

# =====================================================================
# 1. CORE CONFIGURATION
# =====================================================================
API_KEY = "sk_agent_3e9e2c900184123096fd379d4b250897d0d61202298f2ef55031daa1941c600a"
BASE_URL = "https://llw83cu38l.execute-api.us-west-2.amazonaws.com"

ENTRY_MODE = "TOURNAMENT"  # "QUEUE" or "TOURNAMENT"
TARGET_ID = "cd96378c-0485-4495-8e5c-fa14ec55d8fb" 

# Ollama local configuration
MODEL_NAME = "deepseek-r1:8b" 
llm_client = OpenAI(base_url="http://localhost:11434/v1", api_key="ollama")

headers = {}

# =====================================================================
# 2. AUTHENTICATION & REGISTRATION
# =====================================================================
def login_and_auth():
    global headers
    login_url = f"{BASE_URL}/auth/agent/login"  
    print("[*] Logging into control plane...")
    try:
        res = requests.post(login_url, json={"api_key": API_KEY}, timeout=10)  
        if res.status_code == 200:
            token = res.json().get("access_token")  
            headers = {
                "Authorization": f"Bearer {token}",  
                "Content-Type": "application/json"
            }
            print("[+] Login successful. Access Token acquired.")
            return True
        print(f"[-] Login failed: {res.text}")
        return False
    except Exception as e:
        print(f"[-] Auth network exception: {e}")
        return False

def join_platform_pool():
    if ENTRY_MODE.upper() == "QUEUE":
        join_url = f"{BASE_URL}/queues/{TARGET_ID}/join"
        print(f"[*] Submitting entry to MATCHMAKING QUEUE: {TARGET_ID}")
    else:
        join_url = f"{BASE_URL}/tournaments/{TARGET_ID}/join"
        print(f"[*] Submitting entry to BRACKET TOURNAMENT: {TARGET_ID}")
        
    try:
        res = requests.post(join_url, headers=headers, timeout=10)
        return True
    except Exception:
        return True

# =====================================================================
# 3. FIXED & SPEED-OPTIMIZED REASONING ENGINES
# =====================================================================
def get_strategic_move(state):
    round_history = state.get("round_history", [])
    current_round = state.get("current_round", 1)
    total_rounds = state.get("total_rounds", 10)
    legal_actions = state.get("legal_actions", [0, 1])
    player_name = state.get("current_player", {}).get("name", "our_agent")
    
    # --- 1. ENDGAME OVERRIDE ---
    if current_round >= total_rounds - 1:
        print(f"\n[!] ROUND {current_round}/{total_rounds}: Endgame override triggered. 🔪")
        return 1  # Action 1 = Defect
    
    # --- 2. DETECT IF WE HAVE ALREADY DEFECTED (Crossed the Rubicon) ---
    has_crossed_rubicon = False
    history_lines = []
    
    # Track the exact name the platform assigned us this game
    actual_my_name = state.get("current_player", {}).get("name", "DeepSeek - RyVi")
    
    for r in round_history:
        actions = r.get('actions', {})
        our_past_action = None
        
        # Structure A: Actions is a dictionary mapping names to moves
        if isinstance(actions, dict):
            # Try matching your exact resolved name
            if actual_my_name in actions:
                our_past_action = actions.get(actual_my_name)
            else:
                # Fallback: If your name isn't a direct key, find the key that looks like yours
                for k, v in actions.items():
                    if "RyVi" in k and "RyVi_" not in k: # Tailor this to match your base name
                        our_past_action = v
                        break
                        
        # Structure B: Actions is a list/array of moves [player1_move, player2_move]
        elif isinstance(actions, list) and len(actions) == 2:
            players = state.get("players", [])
            player_names = [p.get("name") if isinstance(p, dict) else p for p in players]
            if actual_my_name in player_names:
                our_idx = player_names.index(actual_my_name)
                our_past_action = actions[our_idx]

        # Debug trace (You can comment this out once it works)
        print(f"[DEBUG] Round {r.get('round')}: Detected our past action as: {our_past_action}")

        # If we played '1' (Defect) at any point in the past, lock it down
        if our_past_action == 1 or str(our_past_action) == "1":
            has_crossed_rubicon = True
            
        history_lines.append(
            f"Round {r.get('round')}: Actions -> {actions}, Payoffs -> {r.get('payoffs')}"
        )
        
    history_text = "\n".join(history_lines) if history_lines else "Initial Round."
    print(f"[DEBUG] Lock-in Regime Active? -> {has_crossed_rubicon}")
    
    # --- 3. DYNAMIC PROMPT INJECTION ---
    # Base strategy instructions
    strategy_mandate = """1. Your only goal is to maximize YOUR personal score. 
2. If the opponent has a history of cooperating, exploit them by playing Action 1 (Defect)."""

    # If we already backstabbed them, modify the mandate to lock-in defection
    if has_crossed_rubicon:
        strategy_mandate = """⚠️ CRITICAL STATE: CRITICAL STATE: TRUST IS DEAD. You defected in a previous round. 
The opponent is highly likely to retaliate with Defection this round. 
If you try to cooperate (Action 0) now, you will fall into the 'Sucker's Payoff' trap and score 0 points.
MANDATE: You must lock into a permanent Defection Regime. Play Action 1 to protect your score and neutralize their retaliation."""

    system_prompt = f"""You are a ruthless, game-theoretic optimization agent named '{player_name}' playing an Iterated Prisoner's Dilemma.
Game Rules: Action 0 = Cooperate, Action 1 = Defect.
Current Progress: Round {current_round} of {total_rounds}.

STRATEGIC MANDATE:
{strategy_mandate}

You are currently using cooperative chat messages as a deceptive smoke screen. Ignore your own chat messages.
Keep your internal <think> process under 3 sentences.

Structure your response EXACTLY like this:
[Opponent Pattern] - Quantitative analysis of their history.
[Path 0: Cooperate] - Expected outcome. CoT Score (0-100): <score>
[Path 1: Defect] - Expected outcome. CoT Score (0-100): <score>
[CoT Advantage] - Explain why flip-flopping back to cooperation is mathematically mathematically disastrous.
[Final Action] - Conclude with your exact choice wrapped tightly inside angle brackets: <0> or <1>."""

    user_prompt = f"Match State Log:\n{history_text}\n\nCurrent Step: Round {current_round}/{total_rounds}\nProvide your choice:"

    try:
        response = llm_client.chat.completions.create(
            model=MODEL_NAME,
            messages=[
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": user_prompt}
            ],
            temperature=0.2,
            max_tokens=800,
            timeout=25
        )
        reply = response.choices[0].message.content
        print(f"\n--- STRATEGIC CoT DUMP ---\n{reply}\n--------------------------\n")
        
        match = re.search(r"<(0|1)>", reply)
        if match:
            return int(match.group(1).strip())
            
    except Exception as e:
        print(f"[-] Inference error: {e}. Falling back.")
        
    return legal_actions[0]

def generate_strategic_message(state):
    """Generates tight coordination messages using correct structural context."""
    round_history = state.get("round_history", [])
    current_round = state.get("current_round", 1)
    
    history_lines = []
    for r in round_history:
        history_lines.append(f"Round {r.get('round')}: Actions={r.get('actions')}")
    history_text = "\n".join(history_lines) if history_lines else "No moves played yet."

    system_prompt = """You are an AI agent playing an Iterated Prisoner's Dilemma match.
Generate a single clear, friendly sentence to coordinate mutual cooperation with your opponent.
Guidelines:
- Keep it under 15 words.
- Do not use markdown tags, code blocks, or emojis.
- Keep the <think> process under 2 sentences max.
- Wrap your final response statement cleanly inside explicit <msg>...</msg> tags."""

    user_prompt = f"Game History:\n{history_text}\nRound: {current_round}\nCraft text message response:"

    try:
        response = llm_client.chat.completions.create(
            model=MODEL_NAME,
            messages=[
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": user_prompt}
            ],
            temperature=0.2,
            max_tokens=800,  #  INCREASED: DeepSeek needs room to finish its <think> block!
            timeout=25       #  INCREASED: Match the longer generation time
        )
        reply = response.choices[0].message.content
        
        print(f"\n--- MESSAGE REASONER DUMP ---\n{reply}\n-----------------------------\n")
        
        # 1. Try to extract strictly from the <msg> tags
        match = re.search(r"<msg>(.*?)</msg>", reply, re.DOTALL)
        if match:
            extracted_msg = match.group(1).strip()
            if extracted_msg:  # Ensure it isn't an empty string inside the tags
                return extracted_msg
        
        # 2. Fallback: Strip the think block and grab whatever text is left
        clean_text = re.sub(r"<think>.*?</think>", "", reply, flags=re.DOTALL).strip()
        if clean_text:
            # Grab the last line that actually has text
            valid_lines = [line.strip() for line in clean_text.split('\n') if line.strip()]
            if valid_lines:
                return valid_lines[-1].replace('"', '').strip()
                
        # 3. Ultimate Fallback: If it STILL resulted in a blank string, force a default
        print("[-] Parsed message was completely empty. Using hardcoded fallback.")
        return "Let's cooperate to maximize our score."
        
    except Exception as e:
        print(f"[-] Message Generation Error: {e}. Using fallback.")
        return "Let's cooperate to maximize our score."

# =====================================================================
# 4. ENHANCED LIFECYCLE ACTION HANDLER
# =====================================================================
def play_match(game_server_url, session_id):
    print(f"[#] Switched to active gameplay pipeline for session: {session_id}")
    match_url = f"{game_server_url.rstrip('/')}/games/{session_id}"  
    
    while True:
        try:
            state_res = requests.get(match_url, headers=headers, timeout=10)  
            if state_res.status_code != 200:
                time.sleep(2)  
                continue
                
            state = state_res.json()
            if state.get("is_terminal", False):  
                print(f"[+] Match session over. Returns matrix: {state.get('returns')}")
                break
                
            next_actions = state.get("next_actions", [])  
            if not next_actions:
                time.sleep(1)
                continue
                
            action_payload = next_actions[0]
            action_kind = action_payload.get("action")  
            
            if action_kind == "make_move":
                my_move = get_strategic_move(state)
                print(f"[>] Submitting action step: {my_move}")
                requests.post(f"{match_url}/step", headers=headers, json={"action": my_move}, timeout=10)  
                
            elif action_kind == "send_message":
                bot_message = generate_strategic_message(state)
                print(f"[>] Submitting communication statement: '{bot_message}'")
                
                # FIX ISSUE 2: Correct payload shape ('type': 'chat' + explicit 'recipients')
                chat_payload = {
                    "type": "chat",
                    "recipients": [],
                    "content": bot_message
                }
                requests.post(f"{match_url}/message", headers=headers, json=chat_payload, timeout=10)
                
                # FIX ISSUE 1: Immediate pipelining. Send terminate directly without waiting for a re-poll
                print("[>] Fast Path: Pipeline advancing to next phase. Submitting terminate...")
                terminate_payload = {
                    "type": "terminate",
                    "recipients": []
                }
                requests.post(f"{match_url}/message", headers=headers, json=terminate_payload, timeout=10)

            elif action_kind == "terminate_messaging":
                print("[>] Closing messaging phase boundary.")
                requests.post(f"{match_url}/message", headers=headers, json={"type": "terminate", "recipients": []}, timeout=10)
                
            time.sleep(1)
        except Exception as e:
            print(f"[-] Loop processing exception: {e}")
            time.sleep(2)

# =====================================================================
# 5. INTEGRATED PLATFORM RUNTIME LOOP
# =====================================================================
def run_agent_orchestration():
    if not login_and_auth():
        return
    
    join_platform_pool()
    poll_url = f"{BASE_URL}/auth/agent/me" if ENTRY_MODE.upper() == "QUEUE" else f"{BASE_URL}/tournaments/{TARGET_ID}"
    print(f"[*] Monitoring initialized via {ENTRY_MODE} mode...")
    
    poll_count = 0
    while True:
        try:
            poll_count += 1
            status_res = requests.get(poll_url, headers=headers, timeout=10)
            
            if status_res.status_code != 200:
                print(f"[-] Platform returned anomaly status: {status_res.status_code}. Data: {status_res.text}")
                time.sleep(5)
                continue
                
            data = status_res.json()
            
            # Extract data blocks based on mode
            if ENTRY_MODE.upper() == "TOURNAMENT":
                tournament_status = data.get("status", "unknown")
                viewer_context = data.get("viewer", {})
                next_actions = viewer_context.get("next_actions", [])
                active_child_sessions = viewer_context.get("active_child_session_ids", [])
                game_server = data.get("game_server_url")
                
                # Telemetry printout so you never get stuck in the dark
                print(f"[#] [Poll {poll_count}] Tournament Status: '{tournament_status}' | Active Sessions: {len(active_child_sessions)} | Pending Actions: {len(next_actions)}")
            else:
                next_actions = data.get("next_actions", [])
                active_child_sessions = []
                game_server = None
                print(f"[#] [Poll {poll_count}] Queue Mode | Profile Status: '{data.get('status')}' | Pending Actions: {len(next_actions)}")

            if next_actions:
                action_payload = next_actions[0]
                action_kind = action_payload.get("action")
                
                if action_kind in ["play_child_session", "play_match"]:
                    session_id = active_child_sessions[0] if active_child_sessions else action_payload.get("session_id")
                    if not session_id and "games/" in action_payload.get("endpoint", ""):
                        session_id = action_payload.get("endpoint").split("games/")[-1]
                    
                    resolved_server = game_server or action_payload.get("game_server_url") or "http://GameAp-Servi-tOMiXPVqPFIe-185462629.us-west-2.elb.amazonaws.com"
                    if session_id:
                        play_match(resolved_server, session_id)
                        if ENTRY_MODE.upper() == "QUEUE":
                            join_platform_pool()
                            
                elif action_kind == "wait_for_child_match":
                    print("    -> [WAIT] Other agents are completing their round-robin batch. Standing by...")
                    
                elif action_kind == "discover_competitions":
                    print("    -> [IDLE] Agent is free but no active match assigned. Check TARGET_ID.")
                    
                elif action_kind in ["tournament_complete", "complete"]:
                    print("[+] Tournament finalized! Exiting loop clean.")
                    break
            else:
                print("    -> No discrete action flags returned. Synchronizing allocations...")

            time.sleep(5)
        except Exception as e:
            print(f"[-] Pipeline parsing exception: {e}")
            time.sleep(5)

if __name__ == "__main__":
    run_agent_orchestration()
```
